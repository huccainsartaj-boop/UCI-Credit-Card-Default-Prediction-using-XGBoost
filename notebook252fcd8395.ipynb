```python
# ============================================
# UCI CREDIT CARD DATASET
# PREPROCESSING
# ============================================

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# ============================================
# 1. Load Dataset
# ============================================

file_path = "/kaggle/input/datasets/sartajhuccain/uci-creadit/UCI_Credit_Card.csv"

df = pd.read_csv(file_path)

print("Original Shape:", df.shape)

# ============================================
# 2. Clean Column Names
# ============================================

df.columns = df.columns.str.strip()

# ============================================
# 3. Remove Unnecessary ID Column
# ============================================

if "ID" in df.columns:
    df = df.drop(columns=["ID"])

# ============================================
# 4. Define Target
# ============================================

target = "default.payment.next.month"

# Check target exists
if target not in df.columns:
    raise ValueError(
        f"Target column '{target}' not found.\n"
        f"Available columns: {df.columns.tolist()}"
    )

# ============================================
# 5. Separate Features and Target
# ============================================

X = df.drop(columns=[target])
y = df[target]

print("\nFeatures Shape:", X.shape)
print("Target Shape:", y.shape)

# ============================================
# 6. Handle Missing Values
# ============================================

print("\nMissing Values Before:")
print(X.isnull().sum().sum())

# Numerical columns
numeric_cols = X.select_dtypes(
    include=np.number
).columns

# Fill numerical missing values with median
X[numeric_cols] = X[numeric_cols].fillna(
    X[numeric_cols].median()
)

print("Missing Values After:")
print(X.isnull().sum().sum())

# ============================================
# 7. Convert Data Types
# ============================================

X = X.astype(float)
y = y.astype(int)

# ============================================
# 8. Train-Test Split
# ============================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTraining Data:", X_train.shape)
print("Testing Data:", X_test.shape)

# ============================================
# 9. Feature Scaling
# ============================================

# Scaling is useful for Logistic Regression,
# SVM and KNN.
#
# Decision Tree and XGBoost do NOT require scaling.
#
# Therefore, keep both versions available.

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("\nScaled Training Shape:", X_train_scaled.shape)
print("Scaled Testing Shape:", X_test_scaled.shape)

# ============================================
# 10. Class Distribution
# ============================================

print("\nTarget Distribution:")
print(y.value_counts())

print("\nTarget Percentage:")
print(
    y.value_counts(normalize=True)
      .mul(100)
      .round(2)
)

# ============================================
# 11. Final Check
# ============================================

print("\nFinal Preprocessing Completed!")

print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)

print("\nTarget Classes:")
print(sorted(y.unique()))
```

```python
# ============================================
# XGBOOST MODEL TRAINING
# UCI CREDIT CARD DATASET
# ============================================

from xgboost import XGBClassifier
from sklearn.metrics import accuracy_score, classification_report

# ============================================
# 1. Check Target Classes
# ============================================

print("Target Classes:")
print(sorted(y_train.unique()))

# ============================================
# 2. Create XGBoost Model
# ============================================

model = XGBClassifier(
    n_estimators=200,
    max_depth=6,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    objective="binary:logistic",
    eval_metric="logloss",
    random_state=42,
    n_jobs=-1
)

# ============================================
# 3. Train Model
# ============================================

print("\nTraining XGBoost Model...")

model.fit(
    X_train,
    y_train
)

print("XGBoost Model Training Completed!")

# ============================================
# 4. Prediction
# ============================================

y_pred = model.predict(X_test)

# ============================================
# 5. Accuracy
# ============================================

accuracy = accuracy_score(y_test, y_pred)

print("\nXGBoost Accuracy:")
print(f"{accuracy * 100:.2f}%")

# ============================================
# 6. Classification Report
# ============================================

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        y_pred,
        target_names=["No Default", "Default"]
    )
)
```

```python
# ============================================
# XGBOOST - VISUALIZATION & EVALUATION
# UCI CREDIT CARD DATASET
# ============================================

import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    roc_curve,
    roc_auc_score
)

# ============================================
# 1. Predictions
# ============================================

y_pred = model.predict(X_test)
y_prob = model.predict_proba(X_test)[:, 1]

# ============================================
# 2. Evaluation Metrics
# ============================================

accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, zero_division=0)
recall = recall_score(y_test, y_pred, zero_division=0)
f1 = f1_score(y_test, y_pred, zero_division=0)
auc = roc_auc_score(y_test, y_prob)

print("=" * 60)
print("XGBOOST EVALUATION")
print("=" * 60)

print(f"Accuracy  : {accuracy:.4f} ({accuracy*100:.2f}%)")
print(f"Precision : {precision:.4f}")
print(f"Recall    : {recall:.4f}")
print(f"F1-Score  : {f1:.4f}")
print(f"ROC-AUC   : {auc:.4f}")

# ============================================
# 3. Classification Report
# ============================================

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        y_pred,
        target_names=["No Default", "Default"],
        zero_division=0
    )
)

# ============================================
# 4. Confusion Matrix
# ============================================

cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(7, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["No Default", "Default"],
    yticklabels=["No Default", "Default"]
)

plt.title("XGBoost Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()

# ============================================
# 5. Metrics Visualization
# ============================================

metrics = {
    "Accuracy": accuracy,
    "Precision": precision,
    "Recall": recall,
    "F1-Score": f1,
    "ROC-AUC": auc
}

plt.figure(figsize=(9, 5))

bars = plt.bar(
    metrics.keys(),
    metrics.values()
)

plt.ylim(0, 1)
plt.title("XGBoost Evaluation Metrics")
plt.ylabel("Score")
plt.xticks(rotation=20)

# Add values on bars
for bar, value in zip(bars, metrics.values()):
    plt.text(
        bar.get_x() + bar.get_width()/2,
        value + 0.02,
        f"{value:.3f}",
        ha="center"
    )

plt.show()

# ============================================
# 6. ROC Curve
# ============================================

fpr, tpr, thresholds = roc_curve(
    y_test,
    y_prob
)

plt.figure(figsize=(8, 6))

plt.plot(
    fpr,
    tpr,
    label=f"XGBoost (AUC = {auc:.3f})"
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    label="Random Classifier"
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("XGBoost ROC Curve")
plt.legend()
plt.grid()
plt.show()

# ============================================
# 7. Feature Importance
# ============================================

importance = pd.Series(
    model.feature_importances_,
    index=X_train.columns
).sort_values(ascending=False)

plt.figure(figsize=(10, 8))

importance.head(15).sort_values().plot(
    kind="barh"
)

plt.title("Top 15 XGBoost Feature Importance")
plt.xlabel("Importance")
plt.ylabel("Features")
plt.show()

# ============================================
# 8. Print Top Features
# ============================================

print("\nTop 15 Important Features:")
print(importance.head(15))

print("\n" + "=" * 60)
print("VISUALIZATION & EVALUATION COMPLETED")
print("=" * 60)
```
